```python
from google.colab import files

uploaded = files.upload()
```

```python
file_name = list(uploaded.keys())[0]
print("Uploaded file:", file_name)
```

```python
import pandas as pd

# Load the uploaded file
df = pd.read_csv(file_name)

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Remove "mm" and convert to numbers
df['JUN'] = df['Actual Rainfall: JUN'].str.replace(' mm', '').astype(float)
df['JUL'] = df['Actual Rainfall: JUL'].str.replace(' mm', '').astype(float)

# Scatter plot
sns.scatterplot(x='JUN', y='JUL', data=df)

plt.xlabel('June Rainfall')
plt.ylabel('July Rainfall')
plt.title('June vs July Rainfall')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Remove "mm" and convert to numbers
df['JUN'] = df['Actual Rainfall: JUN'].str.replace(' mm', '').astype(float)
df['JUL'] = df['Actual Rainfall: JUL'].str.replace(' mm', '').astype(float)

# Line plot
plt.plot(df['YEAR'], df['JUN'], label='June')
plt.plot(df['YEAR'], df['JUL'], label='July')

plt.xlabel('Year')
plt.ylabel('Rainfall (mm)')
plt.title('June and July Rainfall')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a bar plot of June Rainfall for the first 10 years
splt_data = df.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x='YEAR', y='JUN', data=splt_data, hue='YEAR', palette='Blues_d', legend=False)

plt.xlabel('Year')
plt.ylabel('June Rainfall (mm)')
plt.title('June Rainfall for First 10 Years (1901-1910)')
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a boxplot for June and July Rainfall
plt.figure(figsize=(8, 6))
sns.boxplot(data=df[['JUN', 'JUL']], palette='Set2')

plt.ylabel('Rainfall (mm)')
plt.title('Distribution of June vs July Rainfall')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a violin plot for June and July Rainfall
plt.figure(figsize=(8, 6))
sns.violinplot(data=df[['JUN', 'JUL']], palette='Pastel1')

plt.ylabel('Rainfall (mm)')
plt.title('Violin Plot of June vs July Rainfall')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate correlation matrix for the numeric columns we have processed
correlation_matrix = df[['YEAR', 'JUN', 'JUL']].corr()

# Set up the matplotlib figure
plt.figure(figsize=(8, 6))

# Draw the heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Heatmap')
plt.show()
```
