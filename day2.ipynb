```python
person = {"name": "Alex", "age": 25, "city": "Mumbai", "Country":"India","height":165}

# Remove last inserted item
last_item = person.popitem()
print("Removed last item:", last_item)
print(person)
last_item = person.popitem()
print("Removed last item:", last_item)
print(person)
```

```python
person = {"name": "Alen", "age": 25, "city": "Mumbai"}
print (person)

# Remove a key and get its value
removed_value = person.pop("age")
print("Removed value:", removed_value)
print(person)
```

```python
person = {"name": "Alen", "age": 25, "city": "Mumbai"}

# Fetch existing key
print(person.get("name"))        # Output: Alex

# Fetch non-existing key
print(person.get("country"))     # Output: None
```
