```python
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5, 6])
arr2 = np.array([[1, 2, 3], [4, 5, 6], [9, 10, 11], [11, 12, 13]])

print("1D Array:", arr1)
print("2D Array:\n", arr2)

zeros = np.zeros((5, 3))
ones = np.ones((2, 3))
identity = np.eye(6)

print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Identity Matrix:\n", identity)

arange_arr = np.arange(1, 46, 5)
linspace_arr = np.linspace(1, 10, 10)

print("Arange:", arange_arr)
print("Linspace:", linspace_arr)

rand_arr = np.random.rand(4, 5)
print("Random Array:\n", rand_arr)
otp = (rand_arr * 1000000).astype(int)
print("Random Array:\n", otp)
```

```python
import numpy as np

matrix = np.arange(1, 13).reshape(4, 3)
print("Matrix:\n", matrix)
print("Transpose:\n", matrix.T)

arr = np.array([10, 20, 30, 40])
print("Original Array:", arr)
print("Array + 5:", arr + 5)
print("Array squared:", arr ** 2)

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("Matrix Multiplication:\n", np.dot(A, B))

unsorted = np.array([3, 1, 4, 1, 5, 9])
print("Sorted Array:", np.sort(unsorted))
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Amit', 'Riya', 'Karan', 'Neha', 'Hritik', 'Shahrukh', 'Sara', 'Dev', 'Mira'],
    'Age': [25, 17, 30, 22, 28, 35, 16, 19, 24],
    'City': ['Pune', 'Mumbai', 'Pune', 'Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai', 'Delhi']
})

print("Single Column:\n", df['Name'])
print("\nMultiple Columns:\n", df[['Name', 'Age']])
print("\nRow by index (iloc):\n", df.iloc[1])
print("\nRow by label (loc):\n", df.loc[1])

print("\nPeople older than 18:\n", df[df['Age'] > 18])

df['Salary'] = [50000, 60000, 55000, 70000, 45000, 90000, 100000, 10000, 20000]
print("\nWith Salary Column:\n", df)

df2 = pd.DataFrame({
    'A': [1, 2, None, 4],
    'B': [5, None, None, 8]
})
print("\nOriginal with NaN:\n", df2)
print("\nDrop NaN:\n", df2.dropna())
print("\nFill NaN with 2.0:\n", df2.fillna(2.0))

grouped = df.groupby('City')['Salary'].mean()
print("\nAverage Salary by City:\n", grouped)

print("\nSorted by Age:\n", df.sort_values(by='Age'))

df_extra = pd.DataFrame({
    'Name': ['Hritik', 'Shahrukh'],
    'Department': ['HR', 'IT']
})
merged = pd.merge(df, df_extra, on='Name', how='outer')
print("\nMerged DataFrame:\n", merged)
```
