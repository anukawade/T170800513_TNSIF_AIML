```python
n = int(input("Enter a number: "))
reverse = 0

while n != 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reverse =", reverse)
```

```python
n = int(input("Enter a number: "))
flag = 0

for i in range(2, n):
    if n % i == 0:
        flag = 1
        break

if flag == 0:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
def sum_digits(n):
    sum = 0

    while n != 0:
        sum = sum + n % 10
        n = n // 10

    return sum


n = int(input("Enter a number: "))

print("Sum of digits =", sum_digits(n))
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(5, 3)
print("Area:", r.area())
print("Perimeter:", r.perimeter())
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, amount):
        self.salary += amount

e = Employee("Riya", 30000)
e.increase_salary(5000)
print(e.name, e.salary)
```

```python
class ShoppingCart:
    def __init__(self, items):
        self.items = items

    def total_bill(self):
        total = 0
        for item, price in self.items.items():
            total += price
        return total

items = {"Book": 250, "Pen": 20, "Bag": 700}

cart = ShoppingCart(items)

for item, price in cart.items.items():
    print(item, ":", price)

print("Total Bill:", cart.total_bill())
```

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)

    def show_books(self):
        print(self.books)

lib = Library()
lib.add_book("Python Basics")
lib.add_book("Data Structures")
lib.remove_book("Python Basics")
lib.show_books()
```
