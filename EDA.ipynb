```python
# ============================================
# TITANIC DATASET - COMPLETE EDA
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# --------------------------------------------
# 1. Load Dataset
# --------------------------------------------

df = pd.read_csv("tested.csv")

print("Dataset loaded successfully!")

# --------------------------------------------
# 2. Basic Information
# --------------------------------------------

print("\n========== FIRST 5 ROWS ==========")
display(df.head())

print("\n========== DATASET SHAPE ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

# --------------------------------------------
# 3. Missing Values BEFORE Cleaning
# --------------------------------------------

print("\n========== MISSING VALUES BEFORE CLEANING ==========")

missing = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing.values,
    "Percentage": (missing.values / len(df) * 100).round(2)
})

display(missing_table)

# --------------------------------------------
# 4. Handle Missing Values
# --------------------------------------------

# Age → Median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Cabin → Unknown
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("\nMissing values handled successfully!")

# --------------------------------------------
# 5. Check Missing Values AFTER Cleaning
# --------------------------------------------

print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# --------------------------------------------
# 6. Check Duplicate Rows
# --------------------------------------------

print("\n========== DUPLICATE ROWS ==========")
print("Duplicate rows:", df.duplicated().sum())

# --------------------------------------------
# 7. Statistical Summary
# --------------------------------------------

print("\n========== STATISTICAL SUMMARY ==========")
display(df.describe())

# --------------------------------------------
# 8. Numerical and Categorical Columns
# --------------------------------------------

numerical_columns = df.select_dtypes(
    include=np.number
).columns

categorical_columns = df.select_dtypes(
    include="object"
).columns

print("\n========== NUMERICAL COLUMNS ==========")
print(list(numerical_columns))

print("\n========== CATEGORICAL COLUMNS ==========")
print(list(categorical_columns))

# ============================================
# VISUAL EDA
# ============================================

# --------------------------------------------
# 9. Age Distribution
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Age"],
    bins=20,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# --------------------------------------------
# 10. Fare Distribution
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    df["Fare"],
    bins=20,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()

# --------------------------------------------
# 11. Survival Count
# --------------------------------------------

if "Survived" in df.columns:

    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=df,
        x="Survived"
    )

    plt.title("Survival Distribution")
    plt.xlabel("Survived")
    plt.ylabel("Number of Passengers")
    plt.show()

# --------------------------------------------
# 12. Gender Distribution
# --------------------------------------------

if "Sex" in df.columns:

    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=df,
        x="Sex"
    )

    plt.title("Gender Distribution")
    plt.xlabel("Gender")
    plt.ylabel("Number of Passengers")
    plt.show()

# --------------------------------------------
# 13. Passenger Class Distribution
# --------------------------------------------

if "Pclass" in df.columns:

    plt.figure(figsize=(7, 5))

    sns.countplot(
        data=df,
        x="Pclass"
    )

    plt.title("Passenger Class Distribution")
    plt.xlabel("Passenger Class")
    plt.ylabel("Number of Passengers")
    plt.show()

# --------------------------------------------
# 14. Age vs Fare Scatter Plot
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare"
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()

# --------------------------------------------
# 15. Age vs Survival
# --------------------------------------------

if "Survived" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x="Survived",
        y="Age"
    )

    plt.title("Age vs Survival")
    plt.xlabel("Survived")
    plt.ylabel("Age")
    plt.show()

# --------------------------------------------
# 16. Gender vs Survival
# --------------------------------------------

if "Survived" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="Sex",
        hue="Survived"
    )

    plt.title("Gender vs Survival")
    plt.xlabel("Gender")
    plt.ylabel("Number of Passengers")
    plt.show()

# --------------------------------------------
# 17. Passenger Class vs Survival
# --------------------------------------------

if "Survived" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="Pclass",
        hue="Survived"
    )

    plt.title("Passenger Class vs Survival")
    plt.xlabel("Passenger Class")
    plt.ylabel("Number of Passengers")
    plt.show()

# --------------------------------------------
# 18. Correlation Heatmap
# --------------------------------------------

plt.figure(figsize=(10, 6))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()

# --------------------------------------------
# 19. Final Dataset
# --------------------------------------------

print("\n========== FINAL DATASET ==========")
display(df.head(10))

print("\nFinal Shape:", df.shape)

print("\n========== EDA COMPLETED ==========")
```
